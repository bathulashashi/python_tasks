```python
from googletrans import Translator
from gtts import gTTS
from playsound import playsound
class translator:
    def __init__(self):
        self.tran=Translator()
    def translate_text(self,text,target_lan):
        result=self.tran.translate(text,dest=target_lan)
        return result.text
def main():
    tran=translator()
    text=input('enter text to convert into another language')
    target_lan=input("Enter target language code (e.g., 'en', 'fr', 'es'): ")
    r=tran.translate_text(text,target_lan)
    tts=gTTS(text=r,lang=target_lan)
    tts.save('tts1.mp3')
    playsound('tts1.mp3')
    

    print(r)
    
main()
    
```
