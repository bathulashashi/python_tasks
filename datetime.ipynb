```python
#add date or sub date for current date
from datetime import date,timedelta
t=date.today()
print(t)
b_10=t+timedelta(days=-10)
a_10=t+timedelta(days=10)
print(b_10)
print(a_10)
```

```python

from datetime import datetime,timedelta
d=datetime.now()
print(d)
t_10=d+timedelta(hours=+10)
print(t_10)
```

```python
from datetime import datetime
t_24="13:30"
t=datetime.strptime(t_24,"%H:%M")
t_12=t.strftime("%I:%M %p")
print(t_12)
```

```python
from datetime import datetime
time_12='1:30 PM'
t=datetime.strptime(time_12,'%I:%M %p')
t_24=t.strftime('%H:%M:%S')
print(t_24)
```
