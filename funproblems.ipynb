```python
def second_largest(lst):
    first = second = float('-inf')
    for num in lst:
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    return second
second_largest([5,7,3,9])
```

```python
def count_case(s):
    upper = lower = 0
    for ch in s:
        if 'A' <= ch <= 'Z':
            upper += 1
        elif 'a' <= ch <= 'z':
            lower += 1
    return upper, lower
count_case('shaShi Bathula')
```

```python
def remove_duplicates(lst):
    result = []
    for i in lst:
        found = False
        for j in result:
            if i == j:
                found = True
                break
        if not found:
            result.append(i)
    return result
remove_duplicates([1,2,3,2,5,3])
```

```python
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False

    count = [0]*256

    for ch in s1:
        count[ord(ch)] += 1
    for ch in s2:
        count[ord(ch)] -= 1

    for i in count:
        if i != 0:
            return False
    return True
is_anagram('shashi','sai')
```

```python
def primes(start, end):
    result = []
    for num in range(start, end+1):
        if num < 2:
            continue
        is_prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            result.append(num)
    return result
primes(2,100)
```

```python
def char_frequency(s):
    freq = {}
    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq
char_frequency('shahsi')
```

```python
def longest_word(sentence):
    words = sentence.split()
    longest = ""
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest
longest_word('my name is bathula shashi chandra')
```

```python
def custom_sort(lst):
    n = len(lst)
    for i in range(n):
        for j in range(0, n-i-1):
            if lst[j] > lst[j+1]:
                temp = lst[j]
                lst[j] = lst[j+1]
                lst[j+1] = temp
    return lst
custom_sort([1,2,39,2,6])
```

```python
def is_armstrong(num):
    temp = num
    count = 0

    # count digits
    while temp > 0:
        count += 1
        temp //= 10

    temp = num
    total = 0

    while temp > 0:
        digit = temp % 10
        total += digit ** count
        temp //= 10

    return total == num
is_armstrong(45)
```

```python
def merge_lists(a, b):
    result = []

    for i in a:
        found = False
        for j in result:
            if i == j:
                found = True
                break
        if not found:
            result.append(i)

    for i in b:
        found = False
        for j in result:
            if i == j:
                found = True
                break
        if not found:
            result.append(i)

    return result
merge_lists([1,2,4,5],[23,45,22])
```
