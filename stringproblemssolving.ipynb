```python
#Reverse a string without using slicing or reverse().
a=input()
b=''
c=len(a)-1
for i in range(0,c+1):
    b+=a[c-i]
print(b)
```

```python
#Check whether a string is a palindrome.
a=input()
b=''
c=len(a)-1
for i in range(0,c+1):
    b+=a[c-i]
if b==a:
    print('palidrome')
else:
    print('not a palidrome')
```

```python
#Find the first non-repeating character in a string.
a=input()
b={}
for i in a:
    if i in b:
        b[i]+=1
    else:
        b[i]=1
for i,j in b.items():
    if j==1:
        print(f'{i}:{j}')
```

```python
# without using int- bulitin function  convert the string into intger  
n=input()
print(type(n))
sign=1
i=0
num=0
if n[0]=='-':
    sign=-1
    i=1
elif n[0]=='+':
    sign=1
    i=1
while i<len(n):
    num=num*10+(ord(n[i])-ord('0'))
    i+=1
num=num*sign
print(num)
print(type(num))
```

```python
#Count the number of vowels and consonants in a string.
n=input()
vowelcount=0
consonants=0
for i in n:
    if i in 'aieouAIEOU':
        vowelcount+=1
    elif i==' ':
        pass
    else:
        consonants+=1
print(f'vowelcount:{vowelcount},consonants:{consonants}')
        
```

```python
#Reverse each word in a string.
a=input().split()
c=len(a)-1
b=''
for i in range(c+1):
    b+=a[c-i]+' '
print(b)
```

```python
#Replace spaces with special characters (e.g., @)
a=input()
b=''
for i in a:
    if i==' ':
        b+='@'
    else:
        b+=i
print(b)
```
