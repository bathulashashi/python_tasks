```python
#without using int-built in function – build a logic to convert string into int
n=input()
print(type(n))
sign=1
i=0
num=0
if n[0]=='-':
    sign=-1
    i=1
elif n[0]=='+':
    sign=1
    i=1
while i<len(n):
    num=num*10+(ord(n[i])-ord('0'))
    i+=1
num=num*sign
print(num)
print(type(num))
```

```python
# Without using title function – build a logic to convert a word into title
n=input()
res=''
word=True
for i in n:
    if i==' ':
        res+=i
        word=True
    elif word :
        if 'a'<=i<='z':
            res+=chr(ord(i)-32)
        else:
            res+=i
        word=False
    else:
        if 'A'<=i<='Z':
            res+=chr(ord(i)+32)
        else:
            res+=i
print(res)
    
```
