```python
#1.	Write a function to calculate the factorial of a number.
def factorial(n):
    fac=1
    for i in range(1,n+1):
        fac*=i
    print(fac)
factorial(3)
```

```python
#2.	Write a function to check if a number is prime
def is_prime(n):
    for i in range(2,n):
        if n%i==0:
            return False
    return True
print(is_prime(6))
```

```python
#3.	Write a function that takes a string and returns the number of vowels.
def vowel():
    a=input()
    count=0
    for i in a:
        if i in 'aieouAIEOU':
            count+=1
    print(count)
vowel()
            
```

```python
#Write a function that returns the nth Fibonacci number. 
def fibonacci():
    a=int(input())
    b,c=0,1
    for i in range(a):
        print(b,end=' ')
        b,c=c,b+c
        
fibonacci()
```

```python
#Create a calculator function that takes two numbers and an operator (+, -, *, /) 
def calcu():
    a=int(input())
    b=int(input())
    c=input('eneter a opertion u need to perform +,-,/,*')
    if c=='+':
        print(a+b)
    elif c=='-':
        print(a-b)
    elif c=='*':
        print(a*b)
    elif c=='/':
        print(a/b)
    else:
        print('invalid operation')
calcu()
```

```python
#Write a function to count the number of words in a sentence.
def counti():
    a=input()
    b=a.split()
    cout=0
    for i in b:
        cout+=1
    print(cout)
counti()
```

```python
# Write a function that checks whether a string is a palindrome
def palid():
    a=input()
    b=''
    c=len(a)-1
    for i in range(0,c+1):
        b+=a[c-i]
    if a==b:
        print('palidroime')
    else:
        print('not a plaidrome')
palid()
```

```python
#Create a function to sort a list without using built-in sort functions.
def sort1(*a):
    b=list(a)
    for i in range(0,len(b)): 
        for j in range(i+1,len(b)):
            if b[i]>b[j]:
                b[i],b[j]=b[j],b[i]
    print(b)
sort1(8,1,7,8,0)
```
