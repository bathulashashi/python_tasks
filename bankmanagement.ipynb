```python
from abc import ABC, abstractmethod
import random
import qrcode
accounts = {}
class bank(ABC):
    @abstractmethod
    def create_account(self):
        pass
    @abstractmethod
    def deposit(self):
        pass
    @abstractmethod
    def withdraw(self):
        pass
    @abstractmethod
    def check_balance(self):
        pass
    @abstractmethod
    def details(self):
        pass

class sbi(bank):
    def create_account(self):
        self.account_no = random.randint(1000000000, 9999999999)
        self.holder_name = input('Enter your full name as per Aadhar: ')
        while True:
            aadhar = input('Enter your 12-digit Aadhar number: ')
            if len(aadhar) == 12 and aadhar.isdigit():
                self.aadhar_no = int(aadhar)
                break
            else:
                print('Enter a valid 12-digit Aadhar number')
        while True:
            mobile = input('Enter your 10-digit mobile number: ')
            if len(mobile) == 10 and mobile.isdigit():
                self.mobile_no = int(mobile)
                break
            else:
                print('Enter a valid 10-digit mobile number')
        self.account_type = input('Select the account type (saving/zero): ').lower()
        self.balance = 0
        if self.account_type == 'saving':
            while True:
                deposit = int(input('Deposit amount for saving account (min 1000): '))
                if deposit >= 1000:
                    self.balance += deposit
                    break
                else:
                    print('Deposit minimum amount of 1000 or above')
        elif self.account_type == 'zero':
            while True:
                deposit = int(input('Deposit amount to open zero account (min 500): '))
                if deposit >= 500:
                    self.balance += deposit
                    break
                else:
                    print('Minimum 500 is required to open an account')
        accounts[self.account_no] = {
            'name': self.holder_name,
            'mobile': self.mobile_no,
            'accounttype': self.account_type,
            'balance': self.balance,
            'aadhar': self.aadhar_no
        }
        

        print('-----ACCOUNT CREATED SUCCESSFULLY--------')
        print(f'Your account number is {self.account_no}')
        print(f'QR code saved as {self.account_no}_details.png')

    def deposit(self):
        account_no = int(input('Enter your account number: '))
        if account_no in accounts:
            amount = float(input('Enter deposit amount: '))
            accounts[account_no]['balance'] += amount
            print('-----DEPOSITED SUCCESSFULLY-------')
        else:
            print('Account not found')

    def withdraw(self):
        account_no = int(input('Enter your account number: '))
        if account_no in accounts:
            amount = float(input('Enter amount to withdraw: '))
            if accounts[account_no]['balance'] >= amount:
                accounts[account_no]['balance'] -= amount
                print('-----WITHDRAWAL SUCCESSFUL----')
            else:
                print('Insufficient balance')
        else:
            print('Enter a correct account number')

    def check_balance(self):
        account_no = int(input('Enter your account number: '))
        if account_no in accounts:
            print(f"Available balance: {accounts[account_no]['balance']}")
        else:
            print('Enter correct account number')

    def details(self):
        account_no = int(input('Enter an account number: '))
        if account_no in accounts:
            print('---ACCOUNT DETAILS---')
            account_details=''
            for k, v in accounts[account_no].items():
                print(f"{k} === {v} \n")
                account_details+=f"{k} === {v} \n"
            img=qrcode.make(account_details)
            display(img)
        else:
            print('Enter correct account number')


# Main program
obj = sbi()
while True:
    print('\n----Select operation to perform------')
    choice = int(input('1. Create account \n2. Deposit \n3. Withdraw \n4. Check balance \n5. Details \n6. Exit\nEnter choice: '))
    if choice == 1:
        obj.create_account()
    elif choice == 2:
        obj.deposit()
    elif choice == 3:
        obj.withdraw()
    elif choice == 4:
        obj.check_balance()
    elif choice == 5:
        obj.details()
    elif choice == 6:
        print('Exiting program...')
        break
    else:
        print('Enter a valid input')
```
