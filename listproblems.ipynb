```python
#check wheather a list conatains unique values or not
k=[1,2,3,4,5,6,7,7]
is_unq=True
for i in range(len(k)):
    for j in range(i+1,len(k)):
        if k[i]==k[j]:
            is_unq=False
            break
        else:
            is_unq=True
if is_unq:
    print('unq')
else:
    print('not unq')
```

```python
#converting a list into ascending order
k=[10,30,13,56,6,5]
for i in range(len(k)):
    for j in range(i+1,len(k)):
        if k[i]>k[j]:
            k[i],k[j]=k[j],k[i]
print(k)
```

```python
#converting a list into descending order
k=[10,30,13,56,6,5]
for i in range(len(k)):
    for j in range(i+1,len(k)):
        if k[i]<k[j]:
            k[i],k[j]=k[j],k[i]
            
print(k)
print(k[1])
```
