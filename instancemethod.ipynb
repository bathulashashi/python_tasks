```python
#instance method
#instance variable is variable stores unique data for each object
#instance method s operate on individual objects.
#The first parameter is usually self

# creating the instance variable inside constructor
class bank:
    def __init__(self):
        self.name='shashi' #instance variable inside constructor
        self.balance=25000  #instance variable inside constructor
obj=bank()
```

```python
#instance varibale inside the method
class room:
    def method1(self):
        self.student='sai' #instance varible inside method
        self.roll=206 # instance variable inside method
obj=room() # creating object
obj.method1() #calling instance method
obj.roll
```

```python
#creatind instance variable outside the class 
class atm:
    def customer(self):
        pass
obj=atm()
obj.customer()
obj.name='shahsi'
print(obj.name)
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

# creating object
s1 = Student("Rahul", 85)

# calling instance method
s1.display()
```
