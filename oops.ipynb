```python
class emp:
    def __init__(self):
        self.empid=101
        self.empname='shashi'
    def method1(self):
        print(self.empid)
class child(emp):
    def method(self):
        print(self.empid)
obj1=emp()
obj1.method1()
obj=child()
obj.method()
print(obj.empid)
```

```python
class student:
    def __init__(self,stid,stage,stname):
        self.stid=stid
        self.stage=stage
        self.stname=stname
obj1=student(1,21,'shashi')
print(obj1.__dict__)
obj2=student(2,22,'hari')
print(obj2.__dict__)
print(obj1.stname)
print(obj2.stname)
```

```python
#instance variable update
#inside class
#outside class
class bank:
    def __init__(self):
        self.balance=10000
    def withdraw(self):
        self.balance+=1000
obj=bank()
obj.balance+=1000
print(obj.balance)
obj.withdraw()
print(obj.balance)
class sbi(bank):
    def update(self):
        self.balance+=5000
        print('updated balance:',self.balance)
obj1=sbi()
obj1.update()
```

```python
#deleting the instance variable
class bank:
    def __init__(self):
        self.name='shashi'
        #del self.name # inside the constructor
    def method(self):
        #del self.name # inside the method
        pass
o=bank()
del o.name # outside the class
print(o.name)
    
```

```python
class bank:
    def __init__(self):
        self.amo=10000
    def luck(self):
        self.amo-=int(input('enter to withdraw'))
        print(self.amo)
obj=bank()
obj.luck()
```

```python
#class method is used to perform opt on class variables
#decatrator is used to init a class variables
#calss variable hold constant values 
class atm:
    @classmethod#inside the class method
    def createacc(cls):
        cls.balance=5000 #class variable
obj=atm()
obj.createacc()
print(obj.balance)
print(atm.__dict__)
```
