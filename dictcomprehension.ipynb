```python
sq = {x: x**2 for x in range(1, 6)}
print(sq)
```

```python
keys = ['a','b','c','d','e']
values = [1, 2, 3, 4, 5]  
d = {k:v for (k,v) in zip(keys, values)}  
print (d)
```

```python
c = {fruit: len(fruit) for fruit in ['apple', 'banana', 'cherry']}
print(c)
```

```python
text = "Tutor Joes"
ascii_values = {char: ord(char) for char in text}
print(text)
print(ascii_values)
```

```python
text = "hello world"
char_frequency = {char: text.count(char) for char in set(text)}
print(text)
print(char_frequency)
```
