```python
#Check whether the given numbers is prime number or not, if it is a prime number,Multiply it with 2.
n=int(input())
for i in range(2,n):
    if n%i==0:
        print('it is not a prime number')
        break
    else:
        print(f'it is a prime number,multiplication of prime number with two:{n*2}')
        break
```

```python
#[10,30,50,20]Arrange the list  into ascending order without using the sort function
a=[10,30,50,20]
for i in range(len(a)):
    for j in  range(i+1,len(a)):
        if a[i]>a[j]:
            a[i],a[j]=a[j],a[i]
print(a) 
```
