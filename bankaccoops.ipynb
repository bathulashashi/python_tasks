```python
#abstarction : hides the data internal implementation and shows only the essential features to users
#abstraction class : combination of abstraction methods
from abc import ABC,abstractmethod
class bank(ABC):
    @abstractmethod
    def create_aaccount(self):
        pass
    @abstractmethod
    def withdraw(self):
        pass
    @abstractmethod
    def check_bal(self):
        pass
class union(bank):
    def create_aaccount(self):
        self.holder='shashi union'
        print(self.holder)
    def withdraw(self):
        print('25000')
    def check_bal(self):
        print('50000')
class sbi(union):
    def create_aaccount(self):
        self.sbiholder='shashi sbi'
        print(self.sbiholder)
    def withdraw(self):
        print('sbi 25000')
    def check_bal(self):
        print('sbi 50000')
obj=union()
obj.create_aaccount()
obj.withdraw()
obj.check_bal()
sbiob=sbi()
sbiob.create_aaccount()
sbiob.withdraw()
sbiob.check_bal()
```
